# Handwritten Digit Recognition — Demo
This notebook loads `best_model.h5` (if available), displays a sample MNIST image, and runs a prediction.

```python
%matplotlib inline
import os
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from keras.models import load_model
from keras.datasets import mnist
```

```python
# Try to load the provided model file
model_path = 'best_model.h5'
model = None
if os.path.exists(model_path):
    try:
        model = load_model(model_path)
        print('Loaded model:', model_path)
    except Exception as e:
        print('Failed loading model:', e)
else:
    print(f'Model file not found at {model_path}.')
```

```python
# Load a sample from MNIST and visualize it
(x_train, y_train), (x_test, y_test) = mnist.load_data()
img = x_test[0]
plt.imshow(img, cmap='gray')
plt.title(f'Ground truth: {y_test[0]}')
plt.axis('off')
plt.show()

# If model is available, preprocess and predict
if model is not None:
    x = img.reshape(1, 28, 28, 1).astype('float32') / 255.0
    preds = model.predict(x)
    print('Predicted probabilities:', preds)
    print('Predicted label:', int(np.argmax(preds, axis=1)[0]))
else:
    print('No model loaded — prediction skipped.')
```

**Notes:** If `best_model.h5` is not present, run `MNIST_Train.ipynb` to train and export the model, or place a compatible Keras `.h5` file in the workspace root.
